Makemore simple Language Model

The idea behind the Makemore language model is that it takes a text file of names/words as input. Based on this input it generates more names/words similar to the input file provided.

We will first start by building a 'Bigram' langauge model. This basically means: Given the character 'x' what character is most likely to follow it.

Step 1: Lets start by visualizing the Bigrams

```python
words = open('names.txt', 'r').read().splitlines()

for w in words[:1]:
    chs = ['.'] +list(w) + ['.']
    for ch1, ch2 in zip(chs, chs[1:]):
        print(ch1,ch2)
```

The '.' is used to signal a start or end of a word. This notation will help us determine which characters are most likely to start and end a word. Now that we have visualized the bigrams, the next step is to establish a counter which keeps track of the most/least frequent bigrams

Step 2: Setup up Bigram Counter

```python
b = {}
for w in words:
    chs = ['.'] +list(w) + ['.']
    for ch1, ch2 in zip(chs, chs[1:]):
        b[(ch1,ch2)]=b.get((ch1,ch2), 0)+1


sorted_bigrams = sorted(b.items(), key=lambda kv: -kv[1])
for pair,count in sorted_bigrams[:20]:
    print(pair,count)
```

Notice that the most frequent bigrams are 'n as an ending', 'a as an ending', 'an', 'a as a start'
